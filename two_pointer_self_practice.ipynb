Reverse Vowels of a String Problem: Given a string, reverse only all the vowels in the string and return it. The vowels are 'a', 'e', 'i', 'o', and 'u', and they can appear in both lower and upper cases, more than once.
Input: "leetcode" Output: "leotcede"

Students Hint: Convert the string to a list first. The left pointer moves right until it finds a vowel. The right pointer moves left until it finds a vowel. When both point to a vowel, swap them!

```python
def reverse_vowel(s):
  updated_s=list(s)
  vowels=set('aeiouAEIOU')
  left=0
  right=len(s)-1
  while left<right:
    if updated_s[left] not in vowels:
      left+=1
    elif updated_s[right] not in vowels:
      right-=1
    else:
      updated_s[left],updated_s[right]=updated_s[right],updated_s[left]
      left+=1
      right-=1
  return ''.join(updated_s)
s='leetcode'
print(reverse_vowel(s))

```

Separate Positives and Negatives Problem: Given an integer array, move all negative numbers to the left side and all positive numbers to the right side in-place. The order of the numbers does not matter.
Input: [-1, 2, -3, 4, 5, -6] Output: [-1, -6, -3, 4, 5, 2] (or any other valid arrangement)

```python
def move_all_negative_to_left_side(s):
  left=0
  right=len(s)-1
  while left<right:
    if s[left]<0:
      left+=1
    elif s[right]>=0:
      right-=1
    else:
      s[left],s[right]=s[right],s[left]
      left+=1
      right-=1
  return s
s=[-1,2,-3,4,5,0,-6]
print(move_all_negative_to_left_side(s))
```

```python
# applying merge sort
def merge_sort(arr):
  n=len(arr)
  if n<=1:
    return arr
  mid=n//2
  left=merge_sort(arr[:mid])
  right=merge_sort(arr[mid:])
  return merged(left,right)
def merged(left,right):
  l=[]
  i=j=0
  while i<len(left) and j<len(right):
    if left[i]<=right[j]:
      l.append(left[i])
      i+=1
    else:
      l.append(right[j])
      j+=1
  l.extend(left[i:])
  l.extend(right[j:])
  return l
arr=[-1,2,-3,4,5,0,-6]
print(merge_sort(arr))

```

Valid Palindrome II (One Deletion allowed) Problem: Given a string s, return True if the string can be a palindrome after deleting at most one character from it.
Input: "abca" Output: True (You could delete the 'c' to get "aba") Students Hint: Start like a normal valid palindrome using two pointers. If you hit a mismatch, you have two choices: skip the left character OR skip the right character. If either resulting substring is a palindrome, return True!

```python
def valid_palindrome(s):
  def check(left,right):
    while left<right:
      if s[left]!=s[right]:
        return False
      left+=1
      right-=1
    return True
  left=0
  right=len(s)-1
  while left<right:
    if s[left]!=s[right]:
      return check(left+1,right) or check(left,right-1)
    left+=1
    right-=1
  return True
s='abba'
print(valid_palindrome(s))



```

Count Pairs Whose Sum is Less than Target Problem: Given an array of integers that is already sorted, and a target integer, return the number of pairs (i, j) where i < j and nums[i] + nums[j] < target.
Input: nums = [-1, 1, 2, 3, 1], target = 2 (Assume sorted: [-1, 1, 1, 2, 3])

Output: 3

```python
def sum_less_than_target(arr,target):
  arr.sort()
  left=0
  right=len(arr)-1
  count=0
  while left<right:
    if arr[left]+arr[right]<target:
      count+=right-left
      left+=1
    else:
      right-=1
  return count
arr=[-1, 1, 2, 3, 1]
target=2
print(sum_less_than_target(arr,target))
```

Container With Most Water Problem: You are given an integer array heights where each element represents the height of a vertical line on a graph. Find two lines that together with the x-axis form a container, such that the container contains the most water. Return the maximum area of water it can store.
Input: [1, 8, 6, 2, 5, 4, 8, 3, 7] Output: 49

```python
def container_with_max_water(arr):
  left=0
  right=len(arr)-1
  most_water=0
  while left<right:
    wide=right-left
    height=min(arr[right],arr[left])
    area=wide*height
    most_water=max(most_water,area)
    if arr[left]<arr[right]:
      left+=1
    else:
      right-=1
  return most_water
arr=[1, 8, 6, 2, 5, 4, 8, 3, 7]
print(container_with_max_water(arr))
```
